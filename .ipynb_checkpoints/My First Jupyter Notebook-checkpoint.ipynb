```python
#Reading in some datasets
#These datasest contain information about shipped goods per calendar year to the ports of Milwaukee and Green Bay
#respectively, with units in the thousands of short tons (i.e. 2 thousand pounds) of cargo. I seperately downloaded 
#the files with information for domestic and foreign shipping, the definitions of which are in the report these 
#numbers are derived from. I have also looked at the same datasets in the report itself. 

#Loading in the datasets for Milwaukee
milwaukee_domestic=pd.read_csv('Milwaukee5YearCargoReport-Domestic.csv')
milwaukee_foreign=pd.read_csv('Milwaukee5YearCargoReport-Foreign.csv')

print("Milwaukee Domestic")
print(milwaukee_domestic.info())
print("Milwaukee Foreign")
print(milwaukee_foreign.info())
```

```python
#Green Bay 
greenbay_domestic=pd.read_csv('Greenbay5YearCargoReport-Domestic.csv')
greenbay_foreign=pd.read_csv('Greenbay5YearCargoReport-Foreign.csv')

print("Green Bay Domestic")
print(greenbay_domestic.info())
print("Green Bay Foreign")
print(greenbay_foreign.info())
```

```python
#Both datasets seem to have the same columns,though the rows are different between the 
#two cities. Notably right away Milwaukee seems to have more rows, so presumably a wider 
#variety of cargo products. 

#Let's view one of the datasets up more close to see if the column names could make more sense
milwaukee_domestic.head()
```

```python
#After looking we see that the first two columns have identifying information for the rows, and the rest are 
#repeating information from 2024 to 2020 across the different shipping categories. We can definitely rename these to 
#clean the datasets more. We could alternatively split these dataframes into to smaller datasets for each shipping category, 
#but this seems unnecessary. 

#Let's develop some standardized column names
top_col_names=milwaukee_domestic.columns
print(top_col_names)
bot_col_names=milwaukee_domestic.iloc[0,:].astype('str')
print(bot_col_names)

#Trying to create new names using for loops 
test_col_names=top_col_names;print(test_col_names)
#Normally I would process with some form of a string search with grep for all the unknowns but I'm going to try 
#something different to test my limitations 

#testing a dictionary for the numbers in column names to label them
test_dict={"Empty":range(0,1)};print(test_dict["Empty"]);0 in test_dict["Empty"];#for i in range(0,2): print(i)

#"0" in test_col_names[0]

#creating a dictionary for the relabeling
top_nyms={"Empty":range(0,2),"All Traffic Directions":range(3,7),"Intraport":range(8,12),"Receipts":range(13,17),"Shipments":range(18,22)}
#print(top_nyms)

top_col_names=top_col_names.str.replace(pat="Unnamed: ",repl='',regex=False)
#print(top_col_names);"0" in [str(i) for i in range(0,1)]

#if "0" in 
print("space")

#for i in range(0,len(top_col_names)): 
#    if top_col_names[i] in [str(j) for j in top_nyms["Empty"]]:
#        print(top_col_names[i])
print(top_col_names)
top_col_names=top_col_names.to_list()
#with a lot of trial and error I realized that top_col_names is a index object instead of a list so we 
#will be changing this otherwise the below loop is not possible 

for i in range(0,len(top_col_names)): 
    if top_col_names[i] in [str(j) for j in top_nyms["Empty"]]:
        print(top_col_names[i]); top_col_names[i]=""
    elif top_col_names[i] in [str(j) for j in top_nyms["All Traffic Directions"]]:
        top_col_names[i]="alltrafficdirections"
    elif top_col_names[i] in [str(j) for j in top_nyms["Intraport"]]:
        top_col_names[i]="intraport"
    elif top_col_names[i] in [str(j) for j in top_nyms["Receipts"]]:
        top_col_names[i]="receipts"
    elif top_col_names[i] in [str(j) for j in top_nyms["Shipments"]]:
        top_col_names[i]="shipments"

print(top_col_names)

#combining For full names and some final cleanup
new_col_names=top_col_names+bot_col_names;new_col_names=new_col_names.to_list()
new_col_names=[j.replace(' ','').replace('CY','').lower() for j in new_col_names ]
print(new_col_names)
```

```python
#Now with our new string we can relable the columns, let's just check that all our columns are the same across the 
#datasets

print("Data frame columns match") 
print(all((milwaukee_domestic.columns == milwaukee_foreign.columns) & (milwaukee_foreign.columns==greenbay_domestic.columns)
    & (greenbay_domestic.columns==greenbay_foreign.columns)))

print("First row columns match")
print(all((milwaukee_domestic.iloc[0,:] == milwaukee_foreign.iloc[0,:]) & (greenbay_domestic.iloc[0,:] == milwaukee_foreign.iloc[0,:])
& (greenbay_domestic.iloc[0,:] == greenbay_foreign.iloc[0,:])))


#adjusting our dataframes, making sure to check our inital one first 
milwaukee_domestic.columns=new_col_names#;milwaukee_domestic.info()
milwaukee_domestic=milwaukee_domestic.iloc[1:,:]#;milwaukee_domestic.head()

milwaukee_foreign.columns=new_col_names;milwaukee_foreign=milwaukee_foreign.iloc[1:,:]
greenbay_foreign.columns=new_col_names;greenbay_foreign=greenbay_foreign.iloc[1:,:]
greenbay_domestic.columns=new_col_names;greenbay_domestic=greenbay_domestic.iloc[1:,:]
greenbay_domestic.head(2)#looks good 

```

```python
#Now that we have a our columns set, we need to check if there's any other ways the data needs to be cleaned

#We could potentially remove the code column of the dataframe by setting the commodity code as the row index,
#this could also then be used as keys for merging the datframes if we were interested
greenbay_domestic.info()
greenbay_domestic.head()

#looking at one of the dataframes, the data types for all columns are listed as strings. We will need to convert 
#these to either integers or floats to analyze our data as well. Then we will see if there are any missing values, 
#though I think since this is technically a census of all the cargo in the ports there should not be any.

#After all this is done we should be ready to compute summary statistics and other information from the datasets

#Note: We will not being any hypothesis testing for this, partly because that's not the point of this project 
#but also because the information we are dealing with is already summary statistics that should be a census of all the 
#shipping done in the two ports. 

selected_cols=[0]+[*range(2,len(milwaukee_foreign.columns))]
milwaukee_domestic.iloc[:,selected_cols]
```

```python
#making our columns floats 
selected_cols=[0]+[*range(2,len(milwaukee_foreign.columns))]

milwaukee_domestic[milwaukee_domestic.columns[selected_cols]]=milwaukee_domestic[milwaukee_domestic.columns[selected_cols]].astype('int64')
milwaukee_foreign[milwaukee_foreign.columns[selected_cols]]=milwaukee_foreign[milwaukee_foreign.columns[selected_cols]].astype('int64')

greenbay_domestic[greenbay_domestic.columns[selected_cols]]=greenbay_domestic[greenbay_domestic.columns[selected_cols]].astype('int64')
greenbay_foreign[greenbay_foreign.columns[selected_cols]]=greenbay_foreign[greenbay_foreign.columns[selected_cols]].astype('int64')

#greenbay_foreign.info()
#greenbay_foreign.head()


#changing our row names 
milwaukee_domestic.index=milwaukee_domestic['code'];milwaukee_domestic=milwaukee_domestic.drop('code',axis=1)
milwaukee_foreign.index=milwaukee_foreign['code'];milwaukee_foreign=milwaukee_foreign.drop('code',axis=1)

greenbay_domestic.index=greenbay_domestic['code'];greenbay_domestic=greenbay_domestic.drop('code',axis=1)
greenbay_foreign.index=greenbay_foreign['code'];greenbay_foreign=greenbay_foreign.drop('code',axis=1)

#greenbay_foreign.head()


#these are all set up now 
```
