```python
#Reading in some datasets
#These datasest contain information about shipped goods to the ports of Milwaukee and Green Bay Respectively, with 
#units in the thousands of short tons (i.e. 2 thousand pounds) of cargo. I seperately downloaded the files 
#with information for domestic and foreign shipping, the definitions of which are in the report these numbers are 
#derived from. 

#Loading in the datasets for Milwaukee
milwaukee_domestic=pd.read_csv('Milwaukee5YearCargoReport-Domestic.csv')
milwaukee_foreign=pd.read_csv('Milwaukee5YearCargoReport-Foreign.csv')

print("Milwaukee Domestic")
print(milwaukee_domestic.info())
print("Milwaukee Foreign")
print(milwaukee_foreign.info())
```

```python
#Green Bay 
greenbay_domestic=pd.read_csv('Greenbay5YearCargoReport-Domestic.csv')
greenbay_foreign=pd.read_csv('Greenbay5YearCargoReport-Foreign.csv')

print("Green Bay Domestic")
print(greenbay_domestic.info())
print("Green Bay Foreign")
print(greenbay_foreign.info())
```
